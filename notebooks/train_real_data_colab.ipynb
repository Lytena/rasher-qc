# Train and evaluate on the real pork-rasher dataset (GPU, e.g. Google Colab)

CPU training on the 3,549-image Version 4 dataset takes many hours; on a free Colab T4 GPU the full pipeline takes roughly 15–25 minutes.

1. *Runtime → Change runtime type → T4 GPU*.
2. Upload the Roboflow **YOLOv8** export zip (Dataset → Version 4 → Download → YOLOv8) when prompted, or mount Google Drive.
3. Run all cells. Download `results.zip` at the end and unzip it into the repository so the results are versioned with the code.

```
# 1. Get the code (replace with your repository URL)
!git clone https://github.com/Lytena/rasher-qc.git
%cd rasher-qc
!pip install -q -e ".[dev]"
```

```
# 2. Get the data: upload the Roboflow zip
from google.colab import files
up = files.upload()                       # choose the downloaded zip
zip_name = next(iter(up))
!mkdir -p data/pork-rasher && unzip -q -o "$zip_name" -d data/pork-rasher
!ls data/pork-rasher
```

```
# 3. COCO-pretrained YOLOv8n weights (used for transfer learning)
!mkdir -p models && wget -q -O models/yolov8n_coco_pretrained.pt https://github.com/ultralytics/assets/releases/download/v8.3.0/yolov8n.pt
```

```
# 4. Full pipeline on the real data (GPU 0). Edit EPOCHS if needed.
!DEVICE=0 EPOCHS=60 bash run_all.sh data/pork-rasher
```

```
# 5. Look at the headline numbers
print(open('results/pork-rasher/SUMMARY.md').read())
```

```
# 6. Bring results back (commit them to the repository afterwards)
!zip -qr results.zip results/pork-rasher models/pork-rasher
files.download('results.zip')
```

**Note on timing numbers:** the latency/roofline results of step 5 describe the machine they were run on. For the real-time question, also run `python scripts/05_performance.py --data data/pork-rasher` on the CPU you would deploy to (e.g. your laptop) and compare.
